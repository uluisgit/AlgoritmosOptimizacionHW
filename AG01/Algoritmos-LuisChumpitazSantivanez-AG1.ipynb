```python
#Sucesión_de_Fibonacci
#https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci
#Calculo del termino n-simo de la suscesión de Fibonacci
def Fibonacci(N:int):
  if N < 2:
    return 1
  else:
    return Fibonacci(N-1)+Fibonacci(N-2)
```

## Devolución de cambio por técnica voraz

```python
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION
```

```python
cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```python
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0
```

## Viaje por el rio. Programación dinámica

```python
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

```python
import math

def obtener_coordenada_eje(punto, indice_eje):
    #Obtener la coordenada de un punto x, y, z
    if isinstance(punto, (int, float)):
        return punto
    return punto[indice_eje]

def calcular_distancia(p1, p2):
    # Cálcular distancia entre dos puntos, basado en el codigo en ClosessPair_Code.cpp sobre "double calDist(Point p1, Point p2)"
    # numeros simples 1D
    if isinstance(p1, (int, float)) and isinstance(p2, (int, float)):
        return abs(p1 - p2)

    # numeros en 2D o 3D
    suma_cuadrados = 0.0
    for i in range(len(p1)):
        diferencia = p1[i] - p2[i]
        suma_cuadrados = suma_cuadrados + (diferencia * diferencia)
        
    return math.sqrt(suma_cuadrados)

def ordenar_puntos_por_eje(lista_puntos, indice_eje):
    # Ordenar los puntos de un eje especifico x, y, z
    # Basado en el codigo en ClosessPair_Code.cpp sobre "vector<Point> mergeSort(vector<Point> vect, string coord)"
    puntos = list(lista_puntos)
    n = len(puntos)
    for i in range(n):
        indice_minimo = i
        for j in range(i + 1, n):
            val_j = obtener_coordenada_eje(puntos[j], indice_eje)
            val_min = obtener_coordenada_eje(puntos[indice_minimo], indice_eje)
            if val_j < val_min:
                indice_minimo = j
        # Intercambio de posicion
        temp = puntos[i]
        puntos[i] = puntos[indice_minimo]
        puntos[indice_minimo] = temp
    return puntos

def filtrar_puntos_por_limite_x(lista_puntos, valor_limite, es_menor_igual):
    # Separacion de puntos
    resultado = []
    for p in lista_puntos:
        coord_x = obtener_coordenada_eje(p, 0)
        if es_menor_igual:
            if coord_x <= valor_limite:
                resultado.append(p)
        else:
            if coord_x > valor_limite:
                resultado.append(p)
    return resultado

def obtener_franja_cercana(lista_puntos, centro_x, distancia_minima):
    # Obener puntos dentro de la franja en referencia al codigo en ClosessPair_Code.cpp sobre "vector<Point> stripPoints;"
    franja = []
    for p in lista_puntos:
        coord_x = obtener_coordenada_eje(p, 0)
        diferencia_x = abs(coord_x - centro_x)
        if diferencia_x < distancia_minima:
            franja.append(p)
    return franja

def closest_pair_brute_force(puntos):
    # fuerza bruta en referencia al codigo en ClosessPair_Code.cpp sobre "vector<Point> bruteForce(vector<Point> p)"
    # asi mismo en revision de la paper paper3-Parallelizable-Algorithm "Algorithm 2: Brute-force 1-NN similarity join."
    n = len(puntos)
    if n < 2:
        return float('inf'), (None, None)
    
    distancia_minima = float('inf')
    mejor_par = (puntos[0], puntos[1])
    
    for i in range(n):
        for j in range(i + 1, n):
            d = calcular_distancia(puntos[i], puntos[j])
            if d < distancia_minima:
                distancia_minima = d
                mejor_par = (puntos[i], puntos[j])
                
    return distancia_minima, mejor_par

def closest_pair_1d_sorted(puntos_ordenados):
    # Resolver 1D considerando que la lista se encuentra ordenada, los 2 puntos mas cercanos.
    distancia_minima = float('inf')
    mejor_par = (None, None)
    
    n = len(puntos_ordenados)
    for i in range(n - 1):
        distancia = puntos_ordenados[i+1] - puntos_ordenados[i]
        if distancia < distancia_minima:
            distancia_minima = distancia
            mejor_par = (puntos_ordenados[i], puntos_ordenados[i+1])
            
    return distancia_minima, mejor_par

def _closest_pair_2d_rec(px, py):
    # Algoritmo recursivo divide y venceras para 2D
    # Basado en el codigo ClosessPair_Code.cpp sobre la funcion "vector<Point> closestPair(vector<Point> points)"
    n = len(px)    
    if n <= 3:
        return closest_pair_brute_force(px)
    
    mitad = n // 2
    punto_medio = px[mitad]
    limite_x = punto_medio[0]
    
    py_izquierda = filtrar_puntos_por_limite_x(py, limite_x, es_menor_igual=True)
    py_derecha = filtrar_puntos_por_limite_x(py, limite_x, es_menor_igual=False)
    
    dist_izq, par_izq = _closest_pair_2d_rec(px[:mitad], py_izquierda)
    dist_der, par_der = _closest_pair_2d_rec(px[mitad:], py_derecha)
    
    if dist_izq < dist_der:
        d_min = dist_izq
        mejor_par = par_izq
    else:
        d_min = dist_der
        mejor_par = par_der
        
    # Creacion de franja, basado en el codigo ClosessPair_Code.cpp sobre "vector<Point> stripPoints;"
    franja = obtener_franja_cercana(py, limite_x, d_min)
    
    tamano_franja = len(franja)
    for i in range(tamano_franja):
        j = i + 1
        while j < tamano_franja:
            diferencia_y = franja[j][1] - franja[i][1]
            if diferencia_y >= d_min:
                break 
            
            d = calcular_distancia(franja[i], franja[j])
            if d < d_min:
                d_min = d
                mejor_par = (franja[i], franja[j])
            j = j + 1
            
    return d_min, mejor_par

def closest_pair_2d(puntos):
    # Llamar a divide y venceras en 2D, basado en el codigo ClosessPair_Code.cpp entre las lineas 313-315
    px = ordenar_puntos_por_eje(puntos, 0) # Ordenar por X
    py = ordenar_puntos_por_eje(puntos, 1) # Ordenar por Y
    return _closest_pair_2d_rec(px, py)

def _closest_pair_3d_rec(px):
    # Basado en el paper "Algorithm 3 (sección Partition Strategy)"
    n = len(px)
    if n <= 3:
        return closest_pair_brute_force(px)
    
    mitad = n // 2
    punto_medio = px[mitad]
    limite_x = punto_medio[0]
    
    dist_izq, par_izq = _closest_pair_3d_rec(px[:mitad])
    dist_der, par_der = _closest_pair_3d_rec(px[mitad:])
    
    if dist_izq < dist_der:
        d_min = dist_izq
        mejor_par = par_izq
    else:
        d_min = dist_der
        mejor_par = par_der
        
    franja = obtener_franja_cercana(px, limite_x, d_min)
    franja_ordenada_y = ordenar_puntos_por_eje(franja, 1)
    
    tamano_franja = len(franja_ordenada_y)
    for i in range(tamano_franja):
        j = i + 1
        while j < tamano_franja:
            diferencia_y = franja_ordenada_y[j][1] - franja_ordenada_y[i][1]
            if diferencia_y >= d_min:
                break
            
            diferencia_z = abs(franja_ordenada_y[j][2] - franja_ordenada_y[i][2])
            if diferencia_z < d_min:
                d = calcular_distancia(franja_ordenada_y[i], franja_ordenada_y[j])
                if d < d_min:
                    d_min = d
                    mejor_par = (franja_ordenada_y[i], franja_ordenada_y[j])
            j = j + 1
            
    return d_min, mejor_par

def closest_pair_3d(puntos):
    # Llamar al algoritmo divide y venceras en 3D
    px = ordenar_puntos_por_eje(puntos, 0)
    return _closest_pair_3d_rec(px)
```
